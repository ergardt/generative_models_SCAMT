```python
# import libraries
import pickle as pi
```

```python
# import the function class
from scripts.model import MolGen
```

```python
# load data
data = []
with open('../data/database_ChEMBL.csv', "r") as f:
    for line in f.readlines()[1:]:
        data.append(line.split("\n")[0])
```

```python
# set the hyperparameters
lr = 1e-3
bs = 512
epochs = 1 
```

```python
# create model
gan_mol = MolGen(data, hidden_dim=128, lr=lr, device="cuda")
```

```python
# create dataloader
loader = gan_mol.create_dataloader(data, batch_size=bs, shuffle=True, num_workers=1)
```

```python
# train model
gan_mol.train_n_steps(loader, max_epoch=epochs, evaluate_every=50)
```

```python
# save model
pi.dump(gan_mol, open('checkpoints/gan_mol_{0}_{1}_{2}k.pkl'.format(bs, lr, epochs), 'wb'))
```
