```python
import pickle as pi
import pandas as pd
```

```python
lr = 1e-3
bs = 512
epochs = 1 

path = 'checkpoints/gan_mol_{0}_{1}_{2}k.pkl'.format(bs, lr, epochs)
```

```python
gan_mol = pi.load(open(path, 'rb'))
```

```python
smiles_list = gan_mol.generate_n(1000)
```

```python
df = pd.DataFrame(smiles_list, columns=["SMILES"])
df.to_csv('result/generated_molcules.csv', index=False)
```
